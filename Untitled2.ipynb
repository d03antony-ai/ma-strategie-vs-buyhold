```python
import yfinance as yf
import pandas as pd
```

```python
ticker = "AAPL"
data = yf.download(ticker, start="2023-01-01", end="2024-01-01")
```

```python
data["MA20"] = data["Close"].rolling(window=20).mean()
data["MA50"] = data["Close"].rolling(window=50).mean()
```

```python
data["Signal"] = 0
data.loc[data["MA20"] > data["MA50"], "Signal"] = 1
data.loc[data["MA20"] < data["MA50"], "Signal"] = -1
```

```python
data["Returns"] = data["Close"].pct_change()
data["Strategy"] = data["Returns"] * data["Signal"].shift(1)
```

```python
cumulative_return = (1 + data["Strategy"]).cumprod().iloc[-1] -1
print(f"Rendement cumulé de la stratégie : {cumulative_return:.2%}")
```

```python
buy_hold_return = (1 + data["Returns"]).cumprod().iloc[-1] -1
```

```python
print(f"Rendement stratégie MA   : {cumulative_return:.2%}")
```

```python
print(f"Rendement Buy & Hold   : {buy_hold_return:.2%}")
```

```python
if cumulative_return > buy_hold_return :
    print("la stratégie bat le Buy & Hold ")
else :
     print(" Le Buy & Hold bat la stratégie ")
```
