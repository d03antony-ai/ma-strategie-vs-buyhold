```python
import pandas as pd
```

```python
def momentum_signal (prices, lockback=20):
    returns = prices.pct_change(lockback)
    signal = pd.Series(0, index=prices.index)
    signal[returns > 0] = 1
    signal[returns < 0] = -1
    return signal

```

```python
prices = pd.Series([100, 102, 101, 105, 113, 117, 120, 118])
signal = momentum_signal(prices)
position = signal.shift(1)
strategy_returns = position * prices.pct_change()
```
